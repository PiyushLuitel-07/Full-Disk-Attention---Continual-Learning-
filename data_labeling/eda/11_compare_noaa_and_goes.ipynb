```python
from pathlib import Path
import pandas as pd

GOES_REFERENCE_CSV = Path(
    "../data_labels/GOES/"
    "NEW_full_dataset_cleaned_1_hours_with_loc_and_time_new.csv"
)

NOAA_CSV = Path(
    "../data_labels/noaa/"
    "labels_noaa_2010_2026.csv"
)
```

```python
goes_reference = pd.read_csv(GOES_REFERENCE_CSV)
noaa = pd.read_csv(NOAA_CSV)
```

```python
print("GOES reference columns:", goes_reference.columns.tolist())
print("NOAA columns:", noaa.columns.tolist())
```

```python
display(goes_reference.head())
display(noaa.head())
```

```python
goes_reference = goes_reference.rename(
    columns={
        "label": "image_path",
        "goes_class": "goes_class_reference",
    }
)
```

```python
noaa = noaa.rename(
    columns={
        "label": "image_path",
        "goes_class": "goes_class_noaa",
    }
)
```

```python
goes_reference["year"] = goes_reference["image_path"].str[:4].astype(int)
noaa["year"] = noaa["image_path"].str[:4].astype(int)
```

```python
goes_reference = goes_reference.query("2010 <= year <= 2018").copy()
noaa = noaa.query("2010 <= year <= 2018").copy()
```

```python
noaa
```

```python
print("GOES reference rows:", len(goes_reference))
print("NOAA rows:", len(noaa))
```

```python
def is_m1_or_larger(goes_class):
    goes_class = str(goes_class).strip().upper()

    if goes_class.startswith("X"):
        return 1

    if goes_class.startswith("M"):
        return int(float(goes_class[1:]) >= 1.0)

    return 0
```

```python
goes_reference["binary_reference"] = (
    goes_reference["goes_class_reference"].apply(is_m1_or_larger)
)
```

```python
goes_reference
```

```python
noaa["binary_noaa"] = (
    noaa["goes_class_noaa"].apply(is_m1_or_larger)
)
```

```python
noaa
```

```python
comparison = goes_reference[
    ["image_path", "goes_class_reference", "binary_reference"]
].merge(
    noaa[
        ["image_path", "goes_class_noaa", "binary_noaa"]
    ],
    on="image_path",
    how="outer",
    indicator=True,
)
```

```python
print(comparison["_merge"].value_counts())
```

compares the two files using image_path as the matching key.
- both: 63,284 — the same image exists in both label files. These are the images we can directly compare.
- left_only: 365 — images only in the GOES-reference file, not in the NOAA file.
- right_only: 25 — images only in the NOAA file, not in the GOES-reference file.
how="outer" keeps every image from both files, so it also reveals missing images rather than silently dropping them

```python
shared = comparison[comparison["_merge"] == "both"].copy()
```

```python
shared
```

```python
shared["exact_goes_class_match"] = (
    shared["goes_class_reference"] == shared["goes_class_noaa"]
)
```

```python
shared
```

```python
shared["binary_label_match"] = (
    shared["binary_reference"] == shared["binary_noaa"]
)
```

```python
shared
```

```python
print(f"Shared images: {len(shared):,}")
print(
    f"Exact GOES-class agreement: "
    f"{shared['exact_goes_class_match'].mean():.2%}"
)
print(
    f"Binary M1+ agreement: "
    f"{shared['binary_label_match'].mean():.2%}"
)
print(
    f"Binary label differences: "
    f"{(~shared['binary_label_match']).sum():,}"
)
```

```python
binary_differences = (
    shared[~shared["binary_label_match"]]
    .sort_values("image_path")
    .reset_index(drop=True)
)
```

```python
display(binary_differences[
    [
        "image_path",
        "goes_class_reference",
        "binary_reference",
        "goes_class_noaa",
        "binary_noaa",
    ]
])
```

- compared the NOAA-based labels with the GOES-reference labels over the 2010–2018 .

- Image coverage is slightly different: 63,284 hourly images are exaclty present in both; 365 occur only in the GOES-reference file and 25 only in the NOAA file. Only the shared images were directly compared.

- For the M1.0+ binary flare target, 63,262 of 63,284 shared images agree: 99.97% agreement, with only 22 differences.

- The 22 differences arise from two catalogue discrepancies: the GOES-reference catalogue contains an M6.4 event beginning at 21:45 on 31 December 2013 that is absent from our NOAA catalogue; and it records an M2.3 event at 23:02 on 9 February 2015, whereas NOAA records a similar M2.4 event at 22:59.
- The small discrepancies arise from catalogue-content/timing differences, not from the labeling code.
